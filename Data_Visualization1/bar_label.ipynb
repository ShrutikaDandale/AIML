```python
import matplotlib.pyplot as plt
```

```python
# bar charts
movies = [
    "the dark knight"
    "the hurt locker"
    "the king's speech"
    "argo"
]
year = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] #in million dollar M$

plt.bar(year, oscar_revenue, color="purple")
plt.title("revenue in each year for oscar movies")
plt.xlabel("year")
plt.ylabel("revenue ()in M$")

# bar label
for i in range(len(year)):
    plt.text(year[i], oscar_revenue[i]+10, str(oscar_revenue[i]), ha="center")

#to extend the size of the chart
plt.ylim(0, max(oscar_revenue)+100)
```
