```python
import matplotlib.pyplot as plt
```

```python
# scatter plots
people = ["Person A", "Person B", "Person C", "Person D", "Person E", ]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]



# customization on scatter plot

# color
# colors = ["red", "pink", "blue", "green", "purple", "red", "pink", "yellow", "green", "purple"]
colors = ["green" if x < 135 else "red" for x in bp]
plt.scatter(age, bp, color=colors, s=bp, marker="o", alpha=0.2) #colour, size=bp, marker=shpe figure, alpha=transparency 
plt.title("age vs bp")
plt.xlabel("age")
plt.ylabel("bp")

plt.grid()
```

```python
# scatter plots
people = ["Person A", "Person B", "Person C", "Person D", "Person E", ]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]



# customization on scatter plot
# colourmap - available on google search for matplotlib colourmqp
plt.scatter(age, bp, s=bp, cmap="plasma", c=bp) #colour, size=bp, marker=shpe figure, alpha=transparency 
plt.title("age vs bp")
plt.xlabel("age")
plt.ylabel("bp")

plt.grid()
plt.colorbar(label="bp")
```
