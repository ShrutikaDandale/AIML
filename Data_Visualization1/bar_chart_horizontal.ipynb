```python
import matplotlib.pyplot as plt
```

```python
# bar charts - horizontal
movies = [
    "the dark knight",
    "the hurt locker",
    "the king's speech",
    "argo",
    "the artist"
]
oscar_revenue = [1005, 170, 427, 133, 232] #in million dollar M$

plt.barh(movies, oscar_revenue) 
plt.title("revenue in each year for oscar movies")
plt.xlabel("revenue")
plt.ylabel("movies")
```
