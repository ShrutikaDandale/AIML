```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
oscar_movies = [
    "the dark knight"
    "the hurt locker"
    "the king's speech"
    "the artist"
    "argo"
]
oscar_revenue = [1005, 170, 427, 133, 232] #in million dollar M$

non_oscar_movies = ["slumdog millionair", "avatat", "inception", "hugo", "lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in M$

year = [2008, 2009, 2010, 2011, 2012]

x = np.arange(len(year))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label="oscar movies") #shifting to left side
plt.bar(x+width/2, non_oscar_revenue, width, label="non-oscar movies") #shifting to right side


plt.title("oscar vs non-oscar movies revenue")
plt.xlabel("year")
plt.ylabel("revenue(in M$)")
plt.legend()
```
