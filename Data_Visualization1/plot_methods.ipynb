```python
import matplotlib.pyplot as plt
```

```python
# Real Data

# Oscar winning movies 
oscar_movies = [
    "the dark knight",
    "the hurt locker",
    "the king's speech",
    "argo"
]
oscar_year = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] #in million dollar M$

# plot
plt.plot(oscar_year, oscar_revenue)

# plot methods
plt.title("oscar movies revenue in each year")
plt.xlabel("years")
plt.ylabel("revenue (in $M)")

plt.tight_layout()

plt.savefig("final_plot.png")
```
