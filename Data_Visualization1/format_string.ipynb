```python
import matplotlib.pyplot as plt
```

```python
# Real Data

# Oscar winning movies 
oscar_movies = [
    "the dark knight"
    "the hurt locker"
    "the king's speech"
    "argo"
]
year = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] #in million dollar M$

# plot
plt.plot(year, oscar_revenue)

# plot methods
plt.title("oscar movies revenue in each year")
plt.xlabel("years")
plt.ylabel("revenue (in $M)")
```

```python
# non oscar winning movies
non_oscar_movies = ["slumdog millionair", "avatat", "inception", "hugo", "lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in M$



# format string
# '[marker][line][color]'

# plt.plot(year, oscar_revenue, "o--r", label="oscar movies")
plt.plot(year, non_oscar_revenue, "<-y", label="non-oscar movies")
plt.plot(year, oscar_revenue, color="red", marker="<", linestyle="-.", linewidth=5, label="oscar movies") #-to change paticullarly the colour, marker of line 


plt.title("oscar movies vs non-oscar movies revenue ")
plt.xlabel("years")
plt.ylabel("revenue (in $M)")


plt.legend()
# plt.legend(loc = "upper left")  - we can also set its location
```
