```python
import matplotlib.pyplot as plt
```

```python
# pie chart- colours and % 
expenses = ["salaries", "rent", "marketing", "R&D", "miscellaneous"]
amounts = [400, 420, 300, 50, 100]

colors=["red", "green", "blue", "yellow", "pink"]

plt.pie(amounts, labels = expenses, autopct="%1.1f%%", colors=colors)
```

```python
# pie chart- wedge
expenses = ["salaries", "rent", "marketing", "R&D", "miscellaneous"]
amounts = [400, 420, 300, 50, 100]

plt.pie(amounts, labels = expenses, autopct="%1.1f%%", 
       wedgeprops= {
           "edgecolor": "black",
           "linewidth": 1,
           "linestyle": "--",
           "alpha": 0.7,
           "fill": False # does not fill any colours
               
       }
       )
plt.show()
```

```python
expenses = ["salaries", "rent", "marketing", "R&D", "miscellaneous"]
amounts = [400, 420, 300, 50, 100]

# explode - use to highlight
explode = [0, 0, 0, 0.2, 0]

plt.pie(amounts, labels = expenses, autopct="%1.1f%%", 
       wedgeprops= {
           "edgecolor": "black",
           "linewidth": 1,
           "linestyle": "--",
           "alpha": 0.9,
       }, explode=explode,
        startangle = 90, # rotate
        shadow=True #3D effect
       )
plt.show()
```
