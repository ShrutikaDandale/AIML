```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

scores = np.random.normal(70, 10, 100) #normal distribution
bins = [0, 30, 70, 90, 100]

plt.hist(scores, color="skyblue", edgecolor="black", bins=bins)
plt.ylabel("# of student")
plt.xlabel("scores")
```
