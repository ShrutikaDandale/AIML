```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# multiple datasets
grp1 = np.random.normal(50, 10, 100)
grp2 = np.random.normal(60, 15, 100)

plt.boxplot([grp1, grp2], tick_labels=["grp1", "grp2"], showmeans=True) 
plt.show()
```

```python
# horizontal box plot
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, vert=False)
plt.grid(True)
```

```python
#showmean- use to show mean
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, vert=False, showmeans=True)
plt.grid(True)
```

```python
# 
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, vert=False, showmeans=True, whis=2.0)
plt.grid(True)
```
