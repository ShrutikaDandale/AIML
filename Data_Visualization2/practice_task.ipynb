```python
import matplotlib.pyplot as plt
```

```python
days = ["mon", "tue", "wed", "thu", "fri"]
cities = ["new york", "london", "delhi", "tokyo"]

temperature = [
    [22, 23, 21, 24, 25],
    [18, 19, 17, 20, 21],
    [30, 32, 31, 33, 34],
    [25, 26, 24, 27, 28]
]

fig, axes = plt.subplots(2, 2)

# 0,0 - 0,1 - 1,0 - 1,1
city_no = 0
for i in range(2): #rows (i) - 0,1
  for j in range(2): # cols(j) - 0,1
    axes[i][j].plot(days, temperature[city_no], marker='o')
    axes[i][j].set_title(cities[city_no])
    axes[i][j].grid(True)
    city_no = city_no + 1

fig.suptitle("temperature in cities over the week")
fig.supylabel("temperature")
fig.supxlabel("days")

fig.tight_layout()
```
