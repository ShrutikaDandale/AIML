```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# scatter plot

tips = sns.load_dataset("tips")

sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip", 
    hue = "time"
)
```

```python
# line plot
flights = sns.load_dataset("flights")

sns.lineplot(
    data = flights,
    x = "year",
    y = "passengers"
)
```
