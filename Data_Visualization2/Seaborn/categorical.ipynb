```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# bar plot

tips = sns.load_dataset("tips")

print(tips.head())
sns.barplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```

```python
# box_plot
sns.boxplot(
    data = tips,
    x = "day",
    y = "tip"
)
```
