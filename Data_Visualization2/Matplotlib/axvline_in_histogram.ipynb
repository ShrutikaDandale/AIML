```python
import matplotlib.pyplot as plt
```

```python
marks = {
    24, 43, 34, 54, 64, 65, 67, 86, 76, 30,
    53, 87, 98, 90, 12, 54, 67, 89, 87, 80,
    23, 54, 67, 87, 89, 13, 14, 54, 67, 87
}

plt.hist(marks, color="skyblue", edgecolor="black", bins=10)

plt.axvline(
    x=33,
    label="passing marks (33)",
    color="red",
    linewidth="2",
    linestyle="--"
)

plt.ylabel("no. of student")
plt.xlabel("marks")
plt.title("distribution marks of students")
```
